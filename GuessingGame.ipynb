```python
#alustetaan arvot "name", "guess" ja "guesses"
name = "Matthew"

guess = ""
guesses = 0


#While-loopin alustus, kunnes arvaus osuu oikeaan.
while guess != name:

    #Kysytaan kayttajaa arvaamaan nimi
    guess = (input("Please, guess my name."))

    #Jokaisesta arvauksesta lisätään "guesses" määrää yhdellä
    guesses += 1


    #Jos arvaus osuu oikeaan, printataan onneksi olkoon ja
    #arvauksien määrä
    if guess == name:
        print("Congratulation!")
        print("You guessed", guesses, "times.")

    #Jos ei osu oikeaan, printataan vihje arvauksien määrän perusteella ja kysytään
    #haluaako käyttäjä lopettaa
    else:
        print("Hint:", name[:guesses])

        quit = input("Do you want to quit (y/n) ?")

        #Jos käyttäjä haluaa lopettaa, loop lopetetaan.
        if quit == "y":
            print("Ok, you can leave if you want.")
            break
```
