```python
import random

thrownDiceNumbers = []

for i in range(5):
    thrownDiceNumbers.append(random.randint(1, 6))

print(thrownDiceNumbers)

print("Sum:", sum(thrownDiceNumbers))

print("Highest value:", max(thrownDiceNumbers))
```
